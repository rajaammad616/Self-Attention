```python
 # Self Attention Implementation in PyTorch
import torch
import torch.nn as nn
from torch.nn import functional as F

class Attention(nn.Module):
    def __init__(self, dim_in, dim_out):
        super().__init__()
        self.dim_in = dim_in
        self.dim_out = dim_out
        self.Q = nn.Linear(dim_in, dim_out)
        self.K = nn.Linear(dim_in, dim_out)
        self.V = nn.Linear(dim_in, dim_out)

    def forward(self, x):
        Q = self.Q(x)
        K = self.K(x)
        V = self.V(x)

        attn_weights = F.softmax(torch.matmul(Q, K.transpose(-2, -1)) / (self.dim_out ** 0.5), dim=-1)
        output = torch.matmul(attn_weights, V)
        return output
```

```python
# Data Creation
import numpy as np
data = np.random.rand(10, 5, 20)  # (batch_size, seq_length, dim_in)
data = torch.tensor(data, dtype=torch.float32) # creating a tensor 
```

```python
# Applying self attention 
attention = Attention(dim_in=20, dim_out=20)
output = attention(data)

print("Input shape:", tuple(data.shape))
print("Attention output shape:", tuple(output.shape))
print("Attention output:")
print(output)
```
